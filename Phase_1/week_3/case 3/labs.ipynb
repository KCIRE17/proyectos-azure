Laboratorio 1: Filtrado de IPs (List Comprehension)

Archivo a crear: laboratorio1.py

Escenario: Te llega una lista de direcciones IP que intentaron acceder a una base de datos en Azure. Debes filtrar y quedarte solo con las IPs que pertenezcan a la red local (aquellas que empiezan exactamente con "192.168").

Tu tarea: Usa List Comprehension (en una sola línea de código) para generar una nueva lista llamada ips_locales e imprímela en la terminal.

```python
log_ips = ["192.168.1.5", "10.0.0.1", "192.168.0.100", "172.16.0.5", "192.168.1.20"]
file = open("filtered_ips.txt", "w")
for ip in log_ips:
    if ip.startswith("192.168."):
        file.write(ip + "\n") 
file.close()    

print("Filtered IP addresses have been written to filtered_ips.txt")
```

Laboratorio 2: Limpieza de Duplicados (Conjuntos / Sets)

Archivo a crear: laboratorio2.py

Escenario: Dos sistemas de ventas distintos registraron clientes el mismo día. Al cruzar la información para subirla al Data Lake, notas que hay IDs de clientes duplicados.

Tu tarea: Une las dos listas en una sola, elimina por completo los duplicados usando la estructura set(), convierte el resultado de vuelta a una lista tradicional y ordénala de menor a mayor antes de imprimirla.

```python
#input
sistema_A = [501, 503, 505, 508]
sistema_B = [502, 503, 508, 510]

#output
union_list = list(set(sistema_A) | set(sistema_B))
list_without_duplicates = set(union_list)
print("The list without duplicates is:", list(list_without_duplicates))
```

Laboratorio 3: El Validador de Esquemas (Diccionarios y JSON)
Archivos a crear: config.json y laboratorio3.py

Escenario: Te llega un archivo de configuración para un pipeline, pero a veces el sistema origen falla y lo envía incompleto. Debes verificar que los datos obligatorios estén presentes.

Tu tarea: 
1. Crea el archivo config.json con el contenido de prueba.

2. En tu script laboratorio3.py, lee ese archivo usando la librería json y corrobora mediante condicionales si las llaves "nombre_tabla" y "columnas" existen dentro del diccionario.

3. Si falta alguna de las dos, muestra un mensaje de error: "Error: Esquema inválido". Si están completas, muestra: "Esquema validado correctamente".

```python
#file: config.json

import json

def validar_esquema():
    try:
        # Leer el archivo config.json
        with open("config.json", 'r') as archivo:
            config = json.load(archivo)
        
        # Corroborar si las llaves existen en el diccionario
        if "nombre_tabla" in config and "columnas" in config:
            print("Esquema validado correctamente")
        else:
            print("Error: Esquema inválido")
            
    except FileNotFoundError:
        print("Error: El archivo config.json no se encuentra en el directorio.")
    except json.JSONDecodeError:
        print("Error: El archivo config.json no tiene un formato JSON válido.")

if __name__ == "__main__":
    validar_esquema()

    
```
