```python
import pandas as pd
import numpy as np

print("Libraries imported successfully.")
```

```python
data = {
    'id_empleado': [101, 102, 103],
    'nombre': ['Ana', 'Bruno', 'Carlos'],
    'salario_usd': [4500, 5200, np.nan] # Un valor nulo (muy común en Ingeniería de Datos)
}

# Convertimos a DataFrame
df = pd.DataFrame(data)
print("DataFrame creado:")
print(df)
```

```python
print("\n--- INFORMACIÓN DE TIPOS ---")
print(df.dtypes)
```

```python
# 1. Filtrar empleados que ganan más de 5000 USD
print("--- EMPLEADOS CON SALARIO MAYOR A 5000 ---")
df_altos_salarios = df[df['salario_usd'] > 5000]
print(df_altos_salarios)

# 2. Detectar filas que tienen valores nulos (NaN)
print("\n--- ¿QUIÉN TIENE VALOR NULO? ---")
df_nulos = df[df['salario_usd'].isna()]
print(df_nulos)
```
