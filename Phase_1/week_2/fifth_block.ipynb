Ejercicio 19: Formateador de Números GrandesEnunciado: 

En ingeniería de datos manejamos billones de registros. Pide al usuario o define 4 números enteros (a, b, c, d) e imprime en pantalla el resultado de la siguiente operación matemática: (a^b) + (c^d). Verás cómo Python maneja números gigantescos automáticamente sin desbordar la memoria (a diferencia de otros lenguajes).

```python
a = int(input("Enter a number (a): "))
b = int(input("Enter another number (b): "))
c = int(input("Enter a third number (c): "))
d = int(input("Enter a fourth number (d): "))

result = (a ** b) + (c ** d)
print(result)
```

Ejercicio 20: El Operador Residuo (Mod)Enunciado: 

Lee o asigna tres números enteros ($a, b, m$).En la primera línea, imprime el resultado de la potencia del primero elevado al segundo ($a^b$).En la segunda línea, imprime el residuo (o resto) de dividir ese resultado de la potencia entre el tercer número $m$ usando el operador de módulo (%).

```python
a = int(input("Enter a number (a): "))
b = int(input("Enter another number (b): "))
m = int(input("Enter a number (m): "))

op = a ** b
print((op))
op_2 = op % m
print(op_2)
```
