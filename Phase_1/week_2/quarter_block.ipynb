Ejercicio 15: El Separador de CSVs (split y join)

Enunciado: Imagina que te llega una línea de un archivo de texto plano: "Erick;25;Data_Engineer;Peru".

Usa el método .split(';') para separar los datos por el punto y coma. Esto te devolverá una lista.

Usa el método .join() para volver a unirlos pero separados por una coma (,), simulando la estructura de un formato CSV estándar. Imprime el resultado final.

```python
file_content = "Erick;25;Data_Engineer;Peru"
file_content.split(";")
msg = ",".join(file_content.split(";"))
print(msg)
```

Ejercicio 16: El f-string de Bienvenida

Enunciado: Crea una función llamada saludar_ingeniero(nombre, apellido) que reciba ambos parámetros. 
Debe retornar exactamente el texto: "¡Hola [nombre] [apellido]! Acabas de ingresar al pipeline de Azure." utilizando la sintaxis de f-strings (f"..."). Prueba la función pasándole tus datos.

```python
def greet_engineer(name,last_name):
    return f"Hello {name} {last_name}, you are into pipeline in Azure!"   

def main():
    name = "Erick"
    last_name = "Ilares Llanos"
    print(greet_engineer(name,last_name))

if __name__ == "__main__":    
    main()
```
