Ejercicio 9: El Segundo Puesto (The Runner-Up)
Enunciado: Tienes una lista de puntajes de un examen: [57, 99, 78, 89, 99, 45, 89]. Escribe un script que encuentre y muestre el segundo puntaje más alto (ignora los duplicados del primer lugar). La respuesta debería ser 89.

```python
list_punt = [57, 99, 78, 89, 99, 45, 89]
punt_unq = sorted(list(set(list_punt)))
print(punt_unq[-2])
```

Ejercicio 10: Notas de Alumnos Anidados
Enunciado: Tienes la siguiente estructura de datos (lista anidada):
alumnos = [["Erick", 9.5], ["Luis", 8.0], ["Ana", 8.0], ["Pedro", 7.0]].
Escribe un código que encuentre al alumno o alumnos con la segunda nota más baja (en este caso, Luis y Ana).

```python
alumnos = [["Erick", 9.5], ["Luis", 8.0], ["Ana", 8.0], ["Pedro", 7.0]]
notas_unicas = sorted(set(alumno[1] for alumno in alumnos))

if len(notas_unicas) > 1:
    segunda_nota_mas_baja = notas_unicas[1]
    alumnos_segunda_nota = [alumno[0] for alumno in alumnos if alumno[1] == segunda_nota_mas_baja]
    
    print(f"La segunda nota más baja es: {segunda_nota_mas_baja}")
    print(f"Alumno(s): {', '.join(alumnos_segunda_nota)}")
else:
    print("No hay suficientes notas diferentes para encontrar una segunda nota más baja.")
```

Ejercicio 11: Buscador en Diccionarios (Data de Clientes)
Enunciado: Crea un diccionario donde las llaves sean nombres de ciudades y los valores sean listas con sus temperaturas de los últimos 3 días:
clima = {"Lima": [22, 24, 23], "Arequipa": [15, 16, 15]}.
Pide al usuario que escriba el nombre de una ciudad y calcula el promedio de sus temperaturas

```python
weather = {"Lima": [22, 24, 23], "Arequipa": [15, 16, 15]}
x = input("Insert the name of the city: ")

if x in weather:
    print(f"The temperatures in {x} is: {sum(weather[x])/len(weather[x])}")
```

Ejercicio 13: Tuplas e Inmutabilidad
Enunciado: Crea una tupla con 5 elementos numéricos. Intenta cambiar el valor del primer elemento mediante código. Captura el error usando un bloque try/except para que el script no se caiga e imprime: "Error: Las tuplas son inmutables".

```python
tuple = (1, 2, 3, 4, 5)

try:
    tuple[0] = 10
    print(tuple)  
except TypeError as e:
    print(f"Error: {e}")
```
