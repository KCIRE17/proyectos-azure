```python
#understanding classes and objects in python
class AzureCard:
    def __init__(self,service,cost_per_month):
        self.service = service
        self.cost_per_month = cost_per_month
    
    def show_info(self):
        print(f"Service: {self.service}, Cost per month: {self.cost_per_month}")

#creating objects and instances
card_1 = AzureCard("Data Factory", 100)
card_2 = AzureCard("Databricks", 50)
card_3 = AzureCard("Azure Synapse", 200)

#accessing attributes and methods
card_1.show_info()
card_2.show_info()
card_3.show_info()
```

```python
class pipeline_ids:
    def __init__(self):
        self.trash_data = [101,None,102,"Error",103,101]
        self.cleaning_data = []
    
    def clean_data(self):
        for_filtrer = []
        for x in self.trash_data:
            if type(x) == int:
                for_filtrer.append(x)

        self.cleaning_data = list(set(for_filtrer))
        print(f"Data cleaned in memory")
    
    def show_report(self):
        print(f"Before data cleaning: {self.trash_data}")
        print(f"After data cleaning: {self.clean_data}")

#executing the pipeline
pipeline = pipeline_ids()
pipeline.clean_data()
pipeline.show_report()  
    
```

```python
import os

# =====================================================================
# 1. CLASE PADRE: El molde base para CUALQUIER pipeline en nuestra empresa
# =====================================================================
class PipelineBase:
    def __init__(self, nombre_fuente):
        self.fuente = nombre_fuente
        # Encapsulamiento: El ID de ejecución es privado (__), nadie lo puede alterar desde fuera
        self.__id_ejecucion = "RUN_9952_AZURE" 

    def iniciar_log(self):
        print(f"[LOG] Iniciando proceso para la fuente: {self.fuente}")
        print(f"[LOG] ID de Auditoría Interna de Azure: {self.__id_ejecucion}")

    # Este método existe aquí pero CADA HIJO lo va a resolver a su manera (Polimorfismo)
    def extraer_datos(self):
        pass


# =====================================================================
# 2. CLASES HIJAS: Heredan de PipelineBase usando (PipelineBase)
# =====================================================================
class PipelineCSV(PipelineBase):
    def extraer_datos(self):
        # Polimorfismo en acción
        print(f"-> Leyendo filas y columnas separadas por comas (,) desde {self.fuente}")
        return [["Erick", "Peru"], ["Luis", "Mexico"]]

class PipelineJSON(PipelineBase):
    def extraer_datos(self):
        # Polimorfismo en acción
        print(f"-> Desarmando estructura de llaves {{}} y valores [] desde {self.fuente}")
        return {"usuario": "KCIRE17", "estado": "Activo"}


# =====================================================================
# 3. EL MUNDO REAL: Ejecutando nuestra arquitectura
# =====================================================================
print("--- EJECUTANDO PIPELINE DE CLIENTES (CSV) ---")
pipeline_clientes = PipelineCSV("clientes_latam.csv")
pipeline_clientes.iniciar_log() # Método heredado del padre
data_csv = pipeline_clientes.extraer_datos() # Método propio del hijo
print(f"Data extraída: {data_csv}\n")

print("--- EJECUTANDO PIPELINE DE CONFIGURACIÓN (JSON) ---")
pipeline_config = PipelineJSON("api_settings.json")
pipeline_config.iniciar_log() # Método heredado del padre
data_json = pipeline_config.extraer_datos() # Método propio del hijo
print(f"Data extraída: {data_json}")
```
